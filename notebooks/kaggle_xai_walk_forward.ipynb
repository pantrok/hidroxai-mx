# HidroXAI-MX · walk-forward anual de las redes en GPU (Kaggle)

Entrena TCN, ConvNeXt-1D y PatchTST en los pliegues anuales del protocolo §3
(entrenamiento 2010 – Y−2, validación Y−1, prueba Y; Y = 2021…2025), con la configuración
elegida en la partición principal y una semilla (especificación en `proof_ledger.md`).
Datos: snapshot v2026.10 (Zenodo 10.5281/zenodo.23150556). Proyecto IND-2026-0335 (IPN, PICDT 2026).

**Antes de correr**
1. *Settings* → *Accelerator*: **GPU T4** (o P100); *Internet*: **On**.
2. *Save Version* → **Save & Run All (Commit)**. Guarda `/kaggle/working/xai_walk_forward` como salida.
3. Si no termina en una sesión: agrega la salida anterior como *Input*, pon su ruta en
   `RESUME_FROM` y vuelve a hacer *Save & Run All*.

Al final, descarga `xai_walk_forward.zip` (pestaña *Output*).


```python
# Parámetros
REPO = "https://github.com/pantrok/hidroxai-mx.git"
COMMIT = "b341ab3fbe17937d103ac31c47150e0d98a9128f"              # código fijo con el que se entrena (commit con scripts/xai_09_walk_forward.py)
ZIP_URL = "https://zenodo.org/records/23150556/files/HidroXAI-MX-v2026.10.zip?download=1"
ZIP_MD5 = "6d0cee07231a747aa0ea307c828ab99d"
RESUME_FROM = None                # p. ej. "/kaggle/input/<salida-anterior>/xai_walk_forward"
OUT = "/kaggle/working/xai_walk_forward"
STOP_AFTER_HOURS = 11.0
N_RUNS = 180                      # 5 años × 4 cuencas × 3 horizontes × 3 redes
```

```python
import hashlib, os, shutil, subprocess, sys, urllib.request, zipfile
import torch
print("GPU:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "ninguna")
assert torch.cuda.is_available(), "Activa la GPU en Settings → Accelerator"

CODE = "/tmp/hidroxai-mx"
if not os.path.exists(CODE):
    subprocess.run(["git", "clone", "-q", REPO, CODE], check=True)
subprocess.run(["git", "-C", CODE, "checkout", "-q", COMMIT], check=True)
print(subprocess.run(["git", "-C", CODE, "log", "-1", "--format=%h %s"], capture_output=True, text=True).stdout)
```

```python
# Dataset v2026.10 desde Zenodo: se verifica el MD5 y se extrae solo lo necesario (fuera de la salida)
DATA = "/tmp/hidromx_data"
ZIP = "/tmp/HidroXAI-MX-v2026.10.zip"
if not os.path.exists(f"{DATA}/features/feature_table.parquet"):
    urllib.request.urlretrieve(ZIP_URL, ZIP)
    md5 = hashlib.md5(open(ZIP, "rb").read()).hexdigest()
    assert md5 == ZIP_MD5, f"MD5 distinto: {md5}"
    with zipfile.ZipFile(ZIP) as zf:
        for m in ("features/feature_table.parquet", "processed/estaciones_hidrorivers.csv"):
            zf.extract(m, DATA)
    os.makedirs(f"{DATA}/raw", exist_ok=True)
    os.remove(ZIP)
print(sorted(os.listdir(DATA)))
```

```python
if RESUME_FROM:
    shutil.copytree(RESUME_FROM, OUT, dirs_exist_ok=True)
    print("Reanudando desde", RESUME_FROM)
os.makedirs(OUT, exist_ok=True)
```

```python
env = {**os.environ, "HIDROXAI_DATA": DATA, "PYTHONPATH": f"{CODE}/src"}
cmd = [sys.executable, "-u", "scripts/xai_09_walk_forward.py", "--models", "tcn", "convnext", "patchtst",
       "--out", OUT, "--device", "cuda", "--stop-after-hours", str(STOP_AFTER_HOURS)]
with subprocess.Popen(cmd, cwd=CODE, env=env, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True) as p:
    for line in p.stdout:
        print(line, end="")
assert p.returncode == 0, f"El script terminó con código {p.returncode}"
```

```python
import pandas as pd
man = pd.read_csv(f"{OUT}/compute_manifest.csv")
n_preds = len(os.listdir(f"{OUT}/preds"))
print(f"Corridas terminadas: {n_preds} de {N_RUNS} · horas de GPU: {man['segundos'].sum() / 3600:.2f}")
shutil.rmtree(f"{OUT}/ckpt", ignore_errors=True) if n_preds == N_RUNS else None
shutil.make_archive("/kaggle/working/xai_walk_forward", "zip", OUT)
print("Salida lista: /kaggle/working/xai_walk_forward.zip")
```
